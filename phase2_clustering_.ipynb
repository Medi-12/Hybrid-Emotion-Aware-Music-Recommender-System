## Phase Objective: Apply Clustering Algorithms

```python
# Import necessary libraries for clustering and dimensionality reduction
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. Fit a K-means clustering model on the genre data using 12 clusters

```python
# Define numerical features for genre clustering
genre_features = ['acousticness', 'danceability', 'duration_ms', 'energy', 'instrumentalness',
                  'liveness', 'loudness', 'speechiness', 'tempo', 'valence', 'popularity']

# Handle potential missing values in genre_data by filling with the mean (or drop if appropriate)
genre_data_numeric = genre_data[genre_features].fillna(genre_data[genre_features].mean())

# Scale the genre features
scaler_genre = StandardScaler()
genre_scaled = scaler_genre.fit_transform(genre_data_numeric)

# Apply K-means clustering with 12 clusters
kmeans_genre = KMeans(n_clusters=7, random_state=42, n_init='auto')
genre_data['genre_cluster'] = kmeans_genre.fit_predict(genre_scaled)

print("Genre clustering complete. First 5 rows with new cluster labels:")
display(genre_data[['genres', 'genre_cluster']].head())
```

### 2. Visualize the clusters of genres using t-SNE dimensionality reduction

```python
# Apply t-SNE for dimensionality reduction on scaled genre data
tsne_genre = TSNE(n_components=2, random_state=42, perplexity=10) # Adjust perplexity if needed
genre_tsne_results = tsne_genre.fit_transform(genre_scaled)

# Create a DataFrame for t-SNE results
genre_tsne_df = pd.DataFrame(data=genre_tsne_results, columns=['tsne_1', 'tsne_2'])
genre_tsne_df['genres'] = genre_data['genres']
genre_tsne_df['cluster'] = genre_data['genre_cluster'].astype(str) # Convert to string for discrete colors

# Plot the genre clusters
fig_genre_tsne = px.scatter(genre_tsne_df,
                            x='tsne_1',
                            y='tsne_2',
                            color='cluster',
                            hover_name='genres',
                            title='Genre Clusters (t-SNE)')
fig_genre_tsne.show()
```
