### 7. Create a 'decade' column in `data` representing the decade of each track

```python
# Ensure 'release_date' is in datetime format, handling potential errors
data['release_date'] = pd.to_datetime(data['release_date'], errors='coerce')

# Filter out rows where release_date could not be parsed (NaT)
data_cleaned = data.dropna(subset=['release_date']).copy()

# Create the decade column
data_cleaned['decade'] = data_cleaned['release_date'].apply(lambda x: f"{x.year // 10 * 10}s")

# Display the first few rows with the new 'decade' column
print('Data with new decade column:')
display(data_cleaned[['release_date', 'decade']].head())
```
